```python
shannonAirport = precipitation[precipitation["Location"] == "Shannon Airport"]
dublinAirport = precipitation[precipitation["Location"] == "Dublin Airport"]
corkAirport = precipitation[precipitation["Location"] == "Cork Airport"]

shannonAirport
```

```python
s1985 = shannonAirport[shannonAirport["Year"] == 1985]
d1985 = dublinAirport[dublinAirport["Year"] == 1985]
c1985 = corkAirport[corkAirport["Year"] == 1985]
```

```python
plt.figure(figsize=(16, 5))
plt.plot(s1985['Month'], s1985["Measurement"], label="Shannon")
plt.plot(d1985['Month'], d1985["Measurement"], label="Dublin")
plt.plot(c1985['Month'], c1985["Measurement"], label="Cork")
plt.legend()
plt.xlabel("Date")
plt.ylabel("Precipitation (mm)")
plt.show()
```
